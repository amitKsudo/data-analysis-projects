## Q7.  Using Titanic dataset

```python
import pandas as pd
print("Amit Kumar\nHIS-23-10\n")
df = pd.read_csv("titanic.csv")
print(df.head())
```

### Q7.a) Find total number of passengers with age less than 30.

```python
print("Amit Kumar\nHIS-23-10\n")
passengersUnder30 = df[df['Age'] < 30].shape[0]
print(f"Total number of pasengers with age less than 30:\n{passengersUnder30}")
```

### Q7.b) Find total fare paid by passengers of first class.

```python
print("Amit Kumar\nHIS-23-10\n")
firstClass = df[df["Pclass"] == 1]
totalFare = firstClass["Fare"].sum()
print(f"Total fair paid by passengers of only first class:\n{totalFare}")
```

### Q7.c) Compare number of survivors of each passenger class.

```python
print("Amit Kumar\nHIS-23-10\n")
survivorClasses = df.groupby('Pclass')['Survived'].sum()
print(f"No. of survivors in each classes:\n{survivorClasses}")
```

### Q7.d) Compute descriptive statistics for any numeric attribute genderwise.

```python
print("Amit Kumar\nHIS-23-10\n")
desStatistics = df.groupby('Sex')['Age'].describe()
print(f"Descriptive statistics for 'Age' attribute, genderwise:\n\n{desStatistics}")
```
